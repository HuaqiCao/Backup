```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

y = np.linspace(-100, 100, 1000)
```

```python

def plot_f_y(a, b):
    f = (-2 * (119.2 - np.sqrt((a - y)**2 + b**2)) * (a - y) * 0.362 / np.sqrt((a - y)**2 + b**2) +
         2 * (119.2 - np.sqrt(y**2 + b**2)) * y * 0.1825 / np.sqrt(y**2 + b**2) +
         2 * (119.2 - np.sqrt((a + y)**2 + b**2)) * (a + y) * 0.362 / np.sqrt((a + y)**2 + b**2) +
         (153.1 - 67 - y) * 0.3843)
    
    plt.rcParams['font.sans-serif'] = ['SimHei']
    plt.rcParams['axes.unicode_minus'] = False

    plt.figure(figsize=(10, 6))
    plt.plot(y, f, linewidth=2, color='blue')
    plt.title(f"f - y图像, a = {a}, b = {b}", fontsize=14)
    plt.xlabel(f'y / mm', fontsize=12)
    plt.ylabel(f'f / N', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()

interact(plot_f_y, 
         a=FloatSlider(min=24, max=44, step=1, value=34, description='a:'), 
         b=FloatSlider(min=55, max=80, step=1, value=67, description='b:'))
```
